# R-MOVE · 05 · ¿Sufren más las incidencias los municipios con menos coches o menos renta?

**Respuesta corta: los municipios con menos coches caen algo más; la renta no marca diferencias. Lo que más protege
sigue siendo tener metro o FGC.**

- **Motorización (evidencia moderada):** a igual distancia y alternativa, +100 turismos por 1.000 habitantes → **1,4 puntos
  menos de caída** en los días de caos (p = 0,008). Se mantiene al quitar cualquier municipio y al controlar por el
  transporte alternativo, pero sin ponderar deja de ser significativo (p = 0,11) y la prueba de permutación queda justo en
  el límite (p = 0,048) (secciones 2 y 3).
- **Renta:** sin efecto en conjunto (p = 0,23). Solo en la avería sin aviso del 13/05/2024 los municipios con más renta
  cayeron más (sección 4).
- **Tener metro o FGC:** ≈ 3 puntos menos de caída (p = 0,008), el factor que más pesa (sección 2).
- **Entre los dependientes**, el tercio con menos coches cae un **−7,1 %** en los días de caos; los otros dos, −1,9 % y
  −3,9 % (sección 5).

## Cómo leer este notebook

**Pregunta.** ¿El impacto de las incidencias de Rodalies en los viajes a Barcelona es distinto según los coches y la renta
de cada municipio?

**Hipótesis.**
- **Motorización:** más coches → menos caída (quien tiene coche puede ir en coche).
- **Renta:** menos renta → menos caída (menos posibilidad de teletrabajar o de no ir).

**Datos.** DGT (turismos y conductores por municipio, 2023–2025), INE Padrón (2023–2025) e INE Atlas de Distribución de
Renta de los Hogares (ADRH, 2023–2024), en `data/raw/socioeco/`. Los viajes, los grupos de municipios y los días de caos
salen de las exportaciones del notebook 03.

**Método.**
1. **Efecto de cada municipio** en los 4 días de caos, con el método del notebook 03 (clustering, corregido por el control).
2. **Regresión ponderada** entre los 62 municipios con Rodalies: efecto ~ motorización + renta + distancia + metro/FGC.
   Cada municipio pesa según su ruido (los municipios pequeños varían mucho de un día a otro).

**Salida.** `observatorio/vulnerabilidad.json` (por municipio) y `observatorio/vulnerabilidad_resumen.json` (modelo,
robustez y terciles), que lee el dashboard.

# A. Preparación de los datos

### A.1 Configuración

```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import statsmodels.formula.api as smf
from scipy.stats import spearmanr

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "data").is_dir())
RAW = ROOT / "data" / "raw" / "socioeco"
PROC = ROOT / "data" / "processed" / "rmove"
sns.set_theme(style="whitegrid")
```

### A.2 Coches por municipio (DGT, 2023–2025)

Excel de la DGT con todos los municipios de España. Se queda la provincia de Barcelona (código INE que empieza por `08`),
con el parque de turismos y el censo de conductores.

```python
dgt = []
for y in [2023, 2024, 2025]:
    d = pd.read_excel(RAW / f"dgt_municipal_{y}.xlsx", dtype={"Código INE": str})
    d = d[d["Código INE"].str[:2] == "08"].assign(anio=y)
    dgt.append(d[["Código INE", "anio", "Parque Turismos", "Censo Conductores"]])

dgt = pd.concat(dgt).rename(columns={"Código INE": "cod", "Parque Turismos": "turismos",
                                     "Censo Conductores": "conductores"})
print(dgt.shape)
dgt.head()
```

### A.3 Población (INE, Padrón 2023–2025)

Es el denominador de la motorización (turismos por 1.000 habitantes). El CSV del INE viene en `latin-1` y usa el punto
como separador de miles.

```python
pad = pd.read_csv(RAW / "padron_08.csv", sep=";", dtype=str, encoding="latin-1")
pad = pad.assign(
    cod=pad.Municipios.str[:5],
    anio=pad.Periodo.astype(int),
    pob=pad.Total.str.replace(".", "", regex=False).astype(float),
)[["cod", "anio", "pob"]]

dgt = dgt.merge(pad, on=["cod", "anio"], how="left")
dgt["tur_1000"] = dgt.turismos / dgt.pob * 1000

print("Sin población:", dgt[dgt.pob.isna()].cod.unique())
dgt.head()
```

```python
tur_anio = dgt.pivot(index="cod", columns="anio", values="tur_1000").drop(index="08000", errors="ignore")

print("Correlación de Spearman entre años:")
display(tur_anio.corr("spearman").round(3))

print("\nMediana por año:")
print(tur_anio.median().round(1))

print("\nMotorización extrema (flotas de empresa registradas en pueblos pequeños):")
display(tur_anio.mean(axis=1).nlargest(5).round(0))
```

**Resultado.** La motorización es muy estable entre años (correlación de Spearman entre 0,98 y 0,99), así que se usa un
único valor por municipio: la media de los tres años. Hay valores extremos por flotas de renting registradas en pueblos
pequeños (Aguilar de Segarra, ≈ 28.800 turismos por 1.000 habitantes), pero ninguno está entre los municipios del análisis.

### A.4 Renta y estructura de los hogares (INE, ADRH 2023–2024)

El fichero mezcla municipios, distritos y secciones censales. Se quedan las filas de municipio (`Distritos` y `Secciones`
vacías) y la media de 2023 y 2024 (2025 aún no está publicado).
- **Renta:** renta neta media por persona y por hogar.
- **Demografía:** tamaño medio del hogar (para pasar a coches por hogar), % de 65 años o más y % de hogares unipersonales.

```python
def leer_adrh(fichero, col_indicador, anios=("2023", "2024")):
    d = pd.read_csv(RAW / fichero, sep=";", dtype=str, encoding="utf-8-sig")
    d = d[d.Distritos.isna() & d.Secciones.isna() & d.Periodo.isin(anios)]   # solo municipios
    d = d.assign(
        cod=d.Municipios.str[:5],
        v=pd.to_numeric(d.Total.str.replace(".", "", regex=False)
                               .str.replace(",", ".", regex=False), errors="coerce"),
    )
    return d.pivot_table(index="cod", columns=col_indicador, values="v", aggfunc="mean")

ren = leer_adrh("adrh_renta_08.csv", "Indicadores de renta media y mediana")
dem = leer_adrh("adrh_demo_08.csv", "Indicadores demográficos")

print("Renta:", ren.shape, "| Demografía:", dem.shape)
print("\nIndicadores de renta:", list(ren.columns))
print("Indicadores demográficos:", list(dem.columns))
```

```python
display(ren[["Renta neta media por persona", "Renta neta media por hogar"]].describe().round(0))
display(dem[["Tamaño medio del hogar", "Porcentaje de población de 65 y más años"]].describe().round(2))
```

### A.5 Tabla socioeconómica por municipio

Un valor por municipio: la media de 2023–2025 para coches y población, y de 2023–2024 para la renta y la demografía.
- `tur_1000`: turismos por 1.000 habitantes (motorización).
- `cond_1000`: conductores con permiso por 1.000 habitantes.
- `tur_hogar`: turismos por hogar (aproximado: `tur_1000` / 1.000 × tamaño medio del hogar).

```python
socio = (dgt[dgt.cod != "08000"]
         .groupby("cod")[["pob", "turismos", "conductores"]].mean())

socio["tur_1000"] = socio.turismos / socio.pob * 1000
socio["cond_1000"] = socio.conductores / socio.pob * 1000

socio = socio.join(
    ren[["Renta neta media por persona", "Renta neta media por hogar"]]
    .set_axis(["renta_pp", "renta_hogar"], axis=1)
)
socio = socio.join(
    dem[["Tamaño medio del hogar", "Porcentaje de población de 65 y más años", "Porcentaje de hogares unipersonales"]]
    .set_axis(["tam_hogar", "pct_65", "pct_unip"], axis=1)
)
socio["tur_hogar"] = socio.tur_1000 / 1000 * socio.tam_hogar

print(socio.shape)
print("Sin renta:", socio.renta_pp.isna().sum(), "municipios")
socio.head()
```

```python
info = pd.read_parquet(PROC / "municipios_info.parquet")
X = info.join(socio)

print("Municipios analizados:", len(X))
print("Sin datos socioeconómicos:", X.tur_1000.isna().sum(), "| sin renta:", X.renta_pp.isna().sum())

display(X.groupby("grupo")[["tur_1000", "tur_hogar", "cond_1000", "renta_pp", "tam_hogar"]].median().round(1))
```

**Resultado.** Los 133 municipios del notebook 03 tienen datos de coches y de renta. Los que tienen metro/FGC son los de
menor motorización (≈ 390 turismos por 1.000 habitantes) y menor renta; los municipios sin Rodalies, los de mayor
motorización (≈ 525). Donde el transporte público es mejor hacen falta menos coches: la motorización está mezclada con
la alternativa de transporte, así que habrá que controlarla en el modelo.

### A.6 Efecto de cada municipio en los días de caos

Para cada municipio *m* y día de caos *d*, el mismo efecto corregido por el control del notebook 03, calculado municipio a
municipio:

$$\text{efecto}_{m,d} = \frac{V_{m,d} \,/\, \text{mediana de } V_m \text{ en los días limpios del mismo tipo y año}}
{V_{\text{control},d} \,/\, \text{lo mismo para la suma de los municipios sin Rodalies}} - 1$$

```python
V = pd.read_parquet(PROC / "municipios_panel_viajes.parquet")              # viajes diarios a BCN: días × municipios
cal = pd.read_parquet(PROC / "municipios_calendario.parquet")
tipo = pd.read_parquet(PROC / "municipios_tipos_dia.parquet").tipo
caos = pd.read_parquet(PROC / "municipios_dias_caos.parquet").fecha.tolist()
ctrl = pd.read_parquet(PROC / "municipios_control.parquet").municipio_code.tolist()

limpios = [d for d in V.index if cal.loc[d, "limpio"]]
V_ctrl = V[ctrl].sum(axis=1)

def efecto_dia(d):
    ref = [x for x in limpios if x.year == d.year and tipo[x] == tipo[d] and x != d]
    ratio_mun = V.loc[d] / V.loc[ref].median()
    ratio_ctrl = V_ctrl.loc[d] / V_ctrl.loc[ref].median()
    return (ratio_mun / ratio_ctrl - 1) * 100

E = pd.DataFrame({d.strftime("%Y-%m-%d"): efecto_dia(d) for d in caos})
E["media_caos"] = E.mean(axis=1)

print("Días de caos:", [d.strftime("%Y-%m-%d") for d in caos], "| tipos:", [tipo[d] for d in caos])
display(info[["grupo"]].join(E).groupby("grupo").median().round(1))
```

**Resultado.** Municipio a municipio se reproduce el patrón del notebook 03: la mediana de los dependientes y de los
cercanos sin metro/FGC cae un ≈ 4,5 % de media en los días de caos, los que tienen metro o FGC apenas cambian (−1,0 %) y
el control queda en ≈ 0. El 26/03/2025 la mediana de los dependientes es positiva: el impacto de esa huelga fue muy
desigual entre municipios.

**Ruido de cada municipio.** Un municipio pequeño puede salir en −15 % solo por azar. Para medir ese ruido se repite el
cálculo tratando cada día limpio como si fuera un día de caos (placebo, efecto real = 0). La variación de la **media de 4
placebos al azar** es el error típico del efecto medio de cada municipio. En la regresión, cada municipio pesa
1 / error²: los más fiables pesan más.

*Simplificación:* a diferencia del notebook 03, aquí los placebos no reentrenan el clustering, así que el error puede
estar algo infraestimado.

```python
PL = pd.DataFrame({d: efecto_dia(d) for d in limpios})          # municipios × días limpios

rng = np.random.default_rng(0)
sims = np.array([PL.iloc[:, rng.choice(PL.shape[1], len(caos), replace=False)].mean(axis=1)
                 for _ in range(2000)])
E["sd_media"] = sims.std(axis=0)

X = info.join(E).join(socio)
print("Error típico de media_caos (pp) por grupo:")
display(X.groupby("grupo").sd_media.describe()[["min", "50%", "max"]].round(1))

# ¿Cuántos municipios con Rodalies caen más de lo que explicaría el azar?
rod = X[X.rodalies]
print("Municipios con Rodalies con caída > 2 errores típicos:",
      (rod.media_caos < -2 * rod.sd_media).sum(), "de", len(rod))
```

**Resultado.** El error típico del efecto medio va de ≈ 1–2 puntos en los municipios grandes a 14–20 puntos en los más
pequeños. 17 de los 62 municipios con Rodalies caen más de 2 errores típicos. Por eso la regresión pondera cada municipio.

Muestra: los 62 municipios con Rodalies.

```python
rod = X[X.rodalies].copy()
rod["log_pob"] = np.log(rod.pob)

etiquetas = {
    "media_caos": "Efecto días de caos",
    "tur_1000": "Turismos / 1.000 hab.",
    "tur_hogar": "Turismos / hogar",
    "cond_1000": "Conductores / 1.000 hab.",
    "renta_pp": "Renta por persona",
    "pct_65": "% ≥ 65 años",
    "dist_km": "Distancia a BCN",
    "log_pob": "Población (log)",
}
corr = rod[list(etiquetas)].corr("spearman").rename(index=etiquetas, columns=etiquetas)

mask = np.triu(np.ones_like(corr, dtype=bool), k=1)   # oculta el triángulo superior (repetido)
fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1,
            center=0, square=True, linewidths=.5, cbar_kws={"shrink": .8, "label": "ρ de Spearman"}, ax=ax)
ax.set_title("Correlaciones entre variables (62 municipios con Rodalies)", pad=12)
plt.tight_layout(); plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(13, 5))
for a, v, lab in [(ax[0], "tur_1000", "Turismos por 1.000 hab."),
                  (ax[1], "renta_pp", "Renta neta media por persona (€)")]:
    sns.scatterplot(data=rod, x=v, y="media_caos", hue="grupo", size="pob",
                    sizes=(20, 400), alpha=.7, ax=a)
    a.axhline(0, c="grey", lw=1)
    a.set_xlabel(lab); a.set_ylabel("Efecto medio en los días de caos (%)")
    r = spearmanr(rod[v], rod.media_caos)
    a.set_title(f"Spearman ρ = {r.statistic:.2f} (p = {r.pvalue:.2f})")
ax[1].get_legend().remove()
ax[0].legend(bbox_to_anchor=(0, -0.15), loc="upper left", ncol=4, frameon=False)
plt.tight_layout(); plt.show()
```

**Resultado.** Sin controles no hay relación entre el efecto y la motorización (ρ = 0,01) ni con la renta (ρ = −0,13). Pero
la motorización está muy ligada al tamaño del municipio (ρ = −0,71) y a la distancia (ρ = 0,35), que a su vez influyen en
el efecto: los municipios con metro/FGC son grandes, tienen pocos coches y apenas caen. Para aislar el papel de los coches
y la renta hay que comparar municipios a igual distancia y con la misma alternativa.

## 2. Regresión ponderada con controles

efecto medio ~ motorización + renta + distancia + metro/FGC

- Pesos: 1 / error² (sección A.6). Errores robustos (HC1).
- Escalas: `tur100` = +100 turismos por 1.000 hab.; `renta_k` = +1.000 € por persona; `dist10` = +10 km.
- Coeficiente **positivo** = **menos caída**.

```python
rod["con_alt"] = rod.con_alt.astype(int)
rod["w"] = 1 / rod.sd_media**2
rod["tur100"] = rod.tur_1000 / 100
rod["renta_k"] = rod.renta_pp / 1000
rod["dist10"] = rod.dist_km / 10

formulas = {
    "Motorización + renta": "media_caos ~ tur100 + renta_k + dist10 + con_alt",
    "Solo motorización":    "media_caos ~ tur100 + dist10 + con_alt",
    "Solo renta":           "media_caos ~ renta_k + dist10 + con_alt",
}
filas = []
for nombre, f in formulas.items():
    m = smf.wls(f, data=rod, weights=rod.w).fit(cov_type="HC1")
    ic = m.conf_int()
    for v in m.params.index.drop("Intercept"):
        filas.append({"modelo": nombre, "variable": v, "coef_pp": m.params[v],
                      "IC95": f"[{ic.loc[v,0]:+.1f}, {ic.loc[v,1]:+.1f}]", "p": m.pvalues[v], "R2": m.rsquared})
res = pd.DataFrame(filas)
display(res.round(3))
```

**Resultado.** Comparando municipios a igual distancia y con la misma alternativa:
- **Motorización:** +100 turismos por 1.000 habitantes → **1,4 puntos menos de caída** (margen de error [+0,4, +2,4],
  p = 0,008). Pasar de 400 a 600 turismos por 1.000 habitantes supone ≈ 3 puntos menos de caída. Apoya la hipótesis de la
  motorización.
- **Renta:** sin efecto significativo (−0,4 puntos por cada +1.000 €, p = 0,23). La hipótesis de la renta no se sostiene.
- **Metro/FGC:** ≈ +3 puntos (p = 0,008): la alternativa de transporte sigue siendo lo que más pesa.
- R² ≈ 0,22: la socioeconomía y el transporte explican una parte modesta de las diferencias entre municipios.

## 3. Robustez del efecto de la motorización

a) Quitar un municipio cada vez. b) Prueba de permutación (repartir la motorización al azar entre municipios).
c) Sin pesos. d) Placebo en los municipios sin Rodalies (no debería salir ninguna relación).

```python
f = "media_caos ~ tur100 + renta_k + dist10 + con_alt"
b0 = smf.wls(f, data=rod, weights=rod.w).fit().params.tur100

# a) Quitar un municipio cada vez
loo = pd.Series({i: smf.wls(f, data=rod.drop(i), weights=rod.drop(i).w).fit().params.tur100 for i in rod.index})
print(f"a) Coef. = {b0:.2f} · quitando un municipio: [{loo.min():.2f}, {loo.max():.2f}]"
      f" · el que más influye: {rod.loc[(loo - b0).abs().idxmax(), 'municipio']}")

# b) Permutación
rng = np.random.default_rng(1)
perm = [smf.wls(f, data=rod.assign(tur100=rng.permutation(rod.tur100.values)), weights=rod.w).fit().params.tur100
        for _ in range(2000)]
print(f"b) p por permutación: {np.mean(np.abs(perm) >= abs(b0)):.3f}")

# c) Sin pesos
m = m_sin_pesos = smf.ols(f, data=rod).fit(cov_type="HC1")
print(f"c) Sin pesos: coef = {m.params.tur100:.2f}, p = {m.pvalues.tur100:.3f}")

# d) Placebo: municipios sin Rodalies
c = X[~X.rodalies].assign(w=lambda d: 1 / d.sd_media**2, tur100=lambda d: d.tur_1000 / 100, renta_k=lambda d: d.renta_pp / 1000)
m = m_placebo = smf.wls("media_caos ~ tur100 + renta_k", data=c, weights=c.w).fit(cov_type="HC1")
print(f"d) Control sin Rodalies: motorización p = {m.pvalues.tur100:.2f}, renta p = {m.pvalues.renta_k:.2f}")
```

**Resultado.**
- **Quitando un municipio cada vez**, el coeficiente se mantiene entre 1,20 y 1,60 (el que más influye es Castellbisbal).
- **Permutación:** p = 0,048, justo por debajo de 0,05.
- **Sin pesos:** coeficiente parecido (1,35), pero deja de ser significativo (p = 0,11).
- **Placebo en los municipios sin Rodalies:** ninguna relación (motorización p = 0,69; renta p = 0,21).

La evidencia es **moderada**: el efecto tiene el tamaño y el signo esperados y no lo produce un solo municipio, pero
depende de dar más peso a los municipios con menos ruido.

## 4. Día a día y control por transporte alternativo

- ¿El patrón se repite en cada día de caos, o lo produce uno solo?
- ¿Se mantiene si se controla también por la conectividad en transporte alternativo (bus interurbano, AMB, FGC), la misma
  que usa el dashboard?

```python
dias = [d.strftime("%Y-%m-%d") for d in caos]
filas = []
for d in dias:
    m = smf.ols(f"Q('{d}') ~ tur100 + renta_k + dist10 + con_alt", data=rod).fit(cov_type="HC1")
    filas.append({"día": d, "coef_motorización": m.params.tur100, "p_motorización": m.pvalues.tur100,
                  "coef_renta": m.params.renta_k, "p_renta": m.pvalues.renta_k})
display(pd.DataFrame(filas).round(3))
```

```python
con = pd.read_parquet(PROC / "conectividad_municipios.parquet")[["sin_alt_directa", "penalizacion", "alternativa"]]
rod = rod.drop(columns=con.columns, errors="ignore").join(con)
rod["sin_alt"] = rod.sin_alt_directa.astype(int)

m_transporte = {}
for nombre, f, datos in [
    ("+ sin alternativa directa", "media_caos ~ tur100 + renta_k + dist10 + con_alt + sin_alt", rod),
    ("+ penalización de tiempo (10 min)", "media_caos ~ tur100 + renta_k + dist10 + con_alt + I(penalizacion/10)",
     rod.dropna(subset=["penalizacion"])),
]:
    m = smf.wls(f, data=datos, weights=datos.w).fit(cov_type="HC1")
    m_transporte[nombre] = m
    extra = m.params.index[-1]
    print(f"{nombre} (n={int(m.nobs)}): motorización {m.params.tur100:.2f} (p={m.pvalues.tur100:.3f}) · "
          f"{extra} {m.params[extra]:.2f} (p={m.pvalues[extra]:.2f})")
```

**Resultado.**
- La motorización tiene signo positivo los 4 días (de +0,6 a +2,3 puntos), pero ninguno es significativo por sí solo: el
  efecto solo se detecta al promediar los días de caos.
- **Renta, solo en la avería sin aviso (13/05/2024):** los municipios con más renta cayeron más (−1,5 puntos por cada
  +1.000 €, p = 0,007), coherente con poder teletrabajar o cancelar ante un imprevisto. En las huelgas anunciadas no
  aparece. Es un único resultado entre 8 pruebas: se plantea como hipótesis.
- Controlando por el transporte alternativo (sin alternativa directa o penalización de tiempo), el efecto de la
  motorización no cambia (1,38 y 1,40; p ≤ 0,016). El bus no añade explicación una vez se tiene en cuenta el metro/FGC.

## 5. Lectura sencilla: los dependientes por terciles

Los municipios dependientes (≥ 20 km, sin metro/FGC) ya son comparables entre sí. Se dividen en tres grupos del mismo
tamaño según la motorización y según la renta, y se calcula el efecto medio de cada grupo (ponderado por el ruido).

```python
dep = rod[rod.grupo == "Dependiente"].copy()
for col, nombre in [("tur_1000", "Motorización (turismos por 1.000 hab.)"), ("renta_pp", "Renta neta por persona (€)")]:
    tercil = pd.qcut(dep[col], 3, labels=["baja", "media", "alta"])
    t = dep.groupby(tercil, observed=True).apply(lambda d: pd.Series({
        "municipios": len(d), "rango": f"{d[col].min():.0f}–{d[col].max():.0f}",
        "efecto_ponderado_%": round(np.average(d.media_caos, weights=d.w), 1)}))
    print(nombre)
    display(t)
```

**Resultado.** El tercio de dependientes con **menos coches** (369–461 turismos por 1.000 habitantes) cae un **−7,1 %**
de media en los días de caos; los de motorización media y alta, −1,9 % y −3,9 %. No es una escalera perfecta (el tercio
intermedio cae menos que el alto), pero el grupo con menos coches es claramente el más afectado. Por renta no hay un patrón:
−3,6 % (baja), −6,6 % (media) y −5,9 % (alta). Es la lectura que muestra el dashboard.

---
## 6. Conclusiones

1. **Motorización, evidencia moderada.** A igual distancia y alternativa, +100 turismos por 1.000 habitantes → 1,4 puntos
   menos de caída (p = 0,008). Se mantiene al quitar municipios y al controlar por el transporte alternativo, y tiene el
   mismo signo los 4 días. Pero sin pesos deja de ser significativo (p = 0,11) y la permutación queda en el límite
   (p = 0,048). Entre los dependientes, el tercio con menos coches es el que más cae (−7,1 %).
2. **Renta, sin efecto en conjunto.** Solo en la avería sin aviso los municipios con más renta cayeron más (hipótesis:
   teletrabajo o cancelación).
3. **Lo que más pesa sigue siendo tener metro o FGC** (≈ 3 puntos menos de caída).

**Limitaciones.** 62 municipios y 4 días de caos (poca potencia); asociación, no causalidad; algunos turismos de empresa se
registran en municipios pequeños; no se sabe qué parte de los viajes de cada municipio se hace en tren.
